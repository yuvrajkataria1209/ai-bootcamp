```python
import numpy as np

# Set a random seed for reproducibility
np.random.seed(42)

students = ["Aman", "Riya", "Karan", "Neha", "Priya", "Mohit", "Sneha", "Arjun", "Meera", "Raj"]

# Generate 10 scores for each subject
maths = np.random.randint(40, 101, 10)
sience = np.random.randint(40, 101, 10)
english = np.random.randint(40, 101, 10)

# Create matrix (10 rows for students, 3 columns for subjects)
matrix = np.array([maths, sience, english]).T

print("STUDENT MARKS")
print("---------------------------------------------------")
print("Student   Maths    Science  English  Total   Average")
print("---------------------------------------------------")

for i in range(10):
    student_total = np.sum(matrix[i,:])
    student_average = np.mean(matrix[i,:])
    print(f"{students[i]:<9} {matrix[i,0]:<8} {matrix[i,1]:<9} {matrix[i,2]:<9} {student_total:<7} {student_average:.2f}")

print("\nMatrix (10x3):")
print(matrix)

print("\nRow-wise Total (Total score for each student):")
print(np.sum(matrix, axis=1))

print("\nColumn-wise Total (Total score for each subject):")
print(np.sum(matrix, axis=0))

print("\nOverall Average:", round(np.mean(matrix), 2))
print("Overall Standard Deviation:", round(np.std(matrix), 2))
```

```python
import matplotlib.pyplot as plt
import numpy as np

# 1. Generate engineering data using NumPy
time = np.linspace(0, 10, 500)  # 500 points from 0 to 10 seconds
amplitude1 = np.sin(2 * np.pi * time)  # Simple sine wave
amplitude2 = np.exp(-0.5 * time) * np.sin(2 * np.pi * time)  # Damped sine wave

# 2. Create a figure with 2 subplots vertically (2 rows, 1 column)
fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(8, 8), dpi=100)

# 3. Plot 1: Simple Sine Wave (Top Subplot)
ax1.plot(
    time,
    amplitude1,
    color="crimson",
    linewidth=2,
    label="Undamped Harmonic Motion",
)
ax1.set_title(
    "Harmonic Response of Structure", fontsize=14, fontweight="bold", pad=12
)
ax1.set_xlabel("Time (seconds)", fontsize=11)
ax1.set_ylabel("Displacement (mm)", fontsize=11)
ax1.grid(True, linestyle="--", alpha=0.6)
ax1.legend(loc="upper right", frameon=True, shadow=True)

# 4. Plot 2: Damped Sine Wave (Bottom Subplot)
ax2.plot(
    time, amplitude2, color="navy", linewidth=2, label="Damped Harmonic Motion"
)
ax2.set_title(
    "Damped Harmonic Response of Structure", fontsize=14, fontweight="bold", pad=12
)
ax2.set_xlabel("Time (seconds)", fontsize=11)
ax2.set_ylabel("Displacement (mm)", fontsize=11)
ax2.grid(True, linestyle="--", alpha=0.6)
ax2.legend(loc="upper right", frameon=True, shadow=True)

# 5. Adjust layout to avoid overlapping labels and display
plt.tight_layout()
# plt.savefig('sensor_data_plot.png', dpi=300, bbox_inches='tight')
plt.show()
```

```python
# Generating Sample Data for Khubh Mela
days = ['Day 1', 'Day 2', 'Day 3', 'Day 4', 'Day 5']
visitors = np.random.randint(50000, 200000, size=5)
cities = ['Prayagraj', 'Varanasi', 'Lucknow', 'Kanpur', 'Agra']
participants = np.random.randint(3000, 15000, size=5)
visitors
#array([122038, 132960,  78229, 108392, 143408], dtype=int32)
# Creating DataFrames
df_visitors = pd.DataFrame({'Day': days, 'Visitors': visitors})
df_participants = pd.DataFrame({'City': cities, 'Participants': participants})
df_visitors
df_participants
# Line Plot: Number of Visitors Over Different Days
plt.figure(figsize=(8, 5))
plt.plot(df_visitors['Day'], df_visitors['Visitors'], marker='o', color='blue')
plt.xlabel('Days')
plt.ylabel('Number of Visitors')
plt.title('Visitors During Khubh Mela Over Days')
plt.grid(True)
plt.show()
```

```python
# Bar Chart: Participants from Different Cities
plt.figure(figsize=(8, 5))
plt.bar(df_participants['City'], df_participants['Participants'], color='orange')
plt.xlabel('Cities')
plt.ylabel('Number of Participants')
plt.title('Participants from Different Cities - Khubh Mela')
plt.show()
```

```python
# Pie Chart: Distribution of Participants by City
plt.figure(figsize=(8, 8))
plt.pie(df_participants['Participants'], labels=df_participants['City'], autopct='%1.1f%%', colors=plt.cm.Paired.colors)
plt.title('Participant Distribution by City')
plt.show()
#colors = plt.cm.Set3.colors --> Soft pastel colors (Great for corporate presentations)
#colors = plt.cm.tab10.colors --> 10 bold, distinct primary colors
#colors = plt.cm.Accent.colors --> High-contrast accent colors
```

```python
# Multiple Plots in One Figure
plt.figure(figsize=(12, 6))

# Subplot 1: Line Plot
plt.subplot(1, 2, 1)
plt.plot(df_visitors['Day'], df_visitors['Visitors'], marker='s', color='green')
plt.title('Daily Visitors')
plt.xlabel('Days')
plt.ylabel('Visitors')

# Subplot 2: Bar Chart
plt.subplot(1, 2, 2)
plt.bar(df_participants['City'], df_participants['Participants'], color='purple')
plt.title('City-wise Participants')
plt.xlabel('Cities')
plt.ylabel('Participants')

plt.tight_layout()
plt.show()

# Adding Legends with Multiple Lines
plt.figure(figsize=(8, 5))
x = np.arange(1, 6)
plt.plot(x, visitors, label='Visitors Trend', linestyle='-', marker='o')
plt.plot(x, participants, label='Participants Trend', linestyle='--', marker='s')
plt.xlabel('Days/Cities Index')
plt.ylabel('Count')
plt.title('Visitors & Participants Trend - Khubh Mela')
plt.legend()
plt.grid(True)
plt.show()

# Saving a Plot
plt.figure(figsize=(8, 5))
plt.plot(df_visitors['Day'], df_visitors['Visitors'], color='red')
plt.title('Saved Plot Example - Visitors Trend')
plt.savefig('khubh_mela_visitors_trend.png')
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Kumbh Mela data
data = {
    "Day": ["Day 1", "Day 2", "Day 3", "Day 4", "Day 5", "Day 6", "Day 7"],
    "Visitors": [50000, 75000, 90000, 120000, 150000, 180000, 200000],
    "City": ["Delhi", "Mumbai", "Delhi", "Lucknow", "Mumbai", "Delhi", "Prayagraj"]
}

df = pd.DataFrame(data)

print(df)

# -------------------------------
# 1. Line Plot - Visitors per Day
# -------------------------------

plt.figure(figsize=(8, 5))

plt.plot(
    df["Day"],
    df["Visitors"],
    marker="o",
    label="Number of Visitors"
)

plt.title("Kumbh Mela Visitors Over Different Days")
plt.xlabel("Days")
plt.ylabel("Number of Visitors")

plt.grid(True)
plt.legend()

plt.savefig("kumbh_mela_visitors.png")
plt.show()


# -------------------------------
# 2. Bar Chart - Participants by City
# -------------------------------

city_counts = df["City"].value_counts()

plt.figure(figsize=(8, 5))

plt.bar(
    city_counts.index,
    city_counts.values,
    label="Participants"
)

plt.title("Distribution of Participants by City")
plt.xlabel("City")
plt.ylabel("Number of Participants")

plt.grid(axis="y")
plt.legend()

plt.savefig("kumbh_mela_cities.png")
plt.show()
```
