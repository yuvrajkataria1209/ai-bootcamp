```python
# Import necessary libraries
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from wordcloud import WordCloud
import matplotlib.pyplot as plt
from collections import Counter

# Ensure necessary NLTK data is downloaded
nltk.download('punkt')
nltk.download('stopwords')
```

```python
text = """Everyone has different hobbies that make their free time enjoyable, and my hobbies are a mixture of creativity, fitness, entertainment, and learning new things. One of my favourite hobbies is **cooking** because I enjoy trying new recipes and experimenting with different ingredients. I find cooking relaxing and satisfying, especially when I get to prepare something for my family or friends. I also enjoy **playing sports and games**, as they keep me active and give me a break from my studies and daily routine. I like spending time outdoors, playing with friends, and participating in activities that require energy and teamwork. **Going to the gym and exercising** is another hobby that I enjoy because fitness is an important part of a healthy lifestyle. Working out helps me stay active, improve my strength, and feel more confident and energetic. Apart from these activities, I love **listening to music** in my free time. Music helps me relax, improves my mood, and makes ordinary activities more enjoyable. I also enjoy **watching movies and series**, especially when I want to take a proper break and entertain myself. Another hobby I have is **travelling and exploring new places**. I like visiting new places, experiencing different surroundings, trying different foods, and creating memories with the people close to me. I am also interested in **photography**, as I enjoy capturing beautiful moments, places, food, and memorable experiences. I have a creative side as well, so I like **designing things, exploring new ideas, and experimenting with technology**, especially when I can create something that looks interesting and useful. I also enjoy **learning new skills** and exploring things that I have never tried before. Sometimes I like spending time on the internet discovering new ideas, tutorials, technologies, or interesting topics. **Dancing** is another activity I enjoy because it is fun, energetic, and a great way to express myself. I also like spending quality time with my friends and family, going out, having conversations, and making memories together. These hobbies not only make my free time enjoyable but also help me maintain a balance between academics and my personal life. They allow me to stay physically active, express my creativity, reduce stress, and continuously explore new interests. Overall, I believe my hobbies reflect my personality because I like trying new things, staying active, being creative, spending time with people I care about, and making the most of my free time.
"""
print(text)
```

```python
nltk.download('punkt_tab')
# Tokenization of words
words = word_tokenize(text.lower())  # Convert to lowercase for uniformity
print(words)
```

```python
# Remove stopwords
stop_words = set(stopwords.words('english'))
filtered_words = [word for word in words if word.isalpha() and word not in stop_words]
# The general syntax for a list comprehension is: [expression for item in iterable if condition]
print(filtered_words)
```

```python
# Create a WordCloud
word_freq = Counter(filtered_words)
wordcloud = WordCloud(width=800, height=400, background_color='white').generate_from_frequencies(word_freq)
```

```python
# Plotting the word cloud
plt.figure(figsize=(10, 5))
plt.imshow(wordcloud, interpolation='bilinear')
plt.axis('off')  # Hide axes
plt.show()
```

```python
from google.colab import files

uploaded = files.upload()

image_name = list(uploaded.keys())[0]

print("Uploaded image:", image_name)
```

```python
import cv2
import matplotlib.pyplot as plt

# Read the uploaded image
image = cv2.imread(image_name)

if image is None:
    print("Error: Image could not be loaded.")
else:
    print("Original Image Shape:", image.shape)

    # Convert BGR to RGB for displaying with Matplotlib
    image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

    # 1. Resize
    resized = cv2.resize(image, (600, 400))
    resized_rgb = cv2.cvtColor(resized, cv2.COLOR_BGR2RGB)

    # 2. Crop
    height, width = image.shape[:2]

    # Safe crop according to image size
    cropped = image[
        height // 4 : 3 * height // 4,
        width // 4 : 3 * width // 4
    ]
    cropped_rgb = cv2.cvtColor(cropped, cv2.COLOR_BGR2RGB)

    # 3. Rotate
    rotated = cv2.rotate(image, cv2.ROTATE_90_CLOCKWISE)
    rotated_rgb = cv2.cvtColor(rotated, cv2.COLOR_BGR2RGB)

    # 4. Flip
    flipped = cv2.flip(image, 1)
    flipped_rgb = cv2.cvtColor(flipped, cv2.COLOR_BGR2RGB)

    # 5. Grayscale
    gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

    # 6. Gaussian Blur
    blurred = cv2.GaussianBlur(image, (7, 7), 0)
    blurred_rgb = cv2.cvtColor(blurred, cv2.COLOR_BGR2RGB)

    # 7. Edge Detection
    edges = cv2.Canny(gray, 100, 200)

    # 8. Brightness and Contrast
    bright_contrast = cv2.convertScaleAbs(
        image,
        alpha=1.3,
        beta=40
    )
    bright_contrast_rgb = cv2.cvtColor(
        bright_contrast,
        cv2.COLOR_BGR2RGB
    )

    # 9. Thresholding
    _, threshold = cv2.threshold(
        gray,
        127,
        255,
        cv2.THRESH_BINARY
    )

    # Save all edited images
    cv2.imwrite("01_resized.jpg", resized)
    cv2.imwrite("02_cropped.jpg", cropped)
    cv2.imwrite("03_rotated.jpg", rotated)
    cv2.imwrite("04_flipped.jpg", flipped)
    cv2.imwrite("05_grayscale.jpg", gray)
    cv2.imwrite("06_blurred.jpg", blurred)
    cv2.imwrite("07_edges.jpg", edges)
    cv2.imwrite("08_brightness_contrast.jpg", bright_contrast)
    cv2.imwrite("09_threshold.jpg", threshold)

    print("All image editing operations completed successfully!")

    # Display results
    plt.figure(figsize=(15, 10))

    plt.subplot(3, 3, 1)
    plt.imshow(image_rgb)
    plt.title("Original")
    plt.axis("off")

    plt.subplot(3, 3, 2)
    plt.imshow(resized_rgb)
    plt.title("Resized")
    plt.axis("off")

    plt.subplot(3, 3, 3)
    plt.imshow(cropped_rgb)
    plt.title("Cropped")
    plt.axis("off")

    plt.subplot(3, 3, 4)
    plt.imshow(rotated_rgb)
    plt.title("Rotated")
    plt.axis("off")

    plt.subplot(3, 3, 5)
    plt.imshow(flipped_rgb)
    plt.title("Flipped")
    plt.axis("off")

    plt.subplot(3, 3, 6)
    plt.imshow(gray, cmap="gray")
    plt.title("Grayscale")
    plt.axis("off")

    plt.subplot(3, 3, 7)
    plt.imshow(blurred_rgb)
    plt.title("Blurred")
    plt.axis("off")

    plt.subplot(3, 3, 8)
    plt.imshow(edges, cmap="gray")
    plt.title("Edge Detection")
    plt.axis("off")

    plt.subplot(3, 3, 9)
    plt.imshow(bright_contrast_rgb)
    plt.title("Brightness + Contrast")
    plt.axis("off")

    plt.tight_layout()
    plt.show()
```
